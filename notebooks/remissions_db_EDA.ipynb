```python
import pandas as pd

years = [2020, 2021, 2022, 2023, 2024, 2025]
dfs = []
for year in years:
    for quarter in ['Q1', 'Q2', 'Q3', 'Q4']:
        file_path = f"../data/raw/BDD PRODUCCION/BDD PRODUCCION/{year}/{quarter} {year}.csv"
        try:
            df = pd.read_csv(file_path)
            dfs.append(df)
        except FileNotFoundError as e:
            print(f"File not found: {file_path}, error: {e}")
            pass 

remissions = pd.concat(dfs, ignore_index=True)
remissions.head(-1)
```

```python
print("Number of rows in remissions DataFrame:", remissions.shape[0])
print("Number of columns in remissions DataFrame:", remissions.shape[1])
```

```python
# Identify rows where both 'ship_addr_line' and 'map_page' are null
remissions[(remissions['ship_addr_line'].isnull() == True) & (remissions['map_page'].isnull() == True)]
```

```python
#Null checking
remissions_percentage_null = remissions.isnull().mean() * 100
print(remissions_percentage_null)
```

```python
#Duplicate checking, where tkc_code can be different
cols_to_check = [col for col in remissions.columns if col != 'tkt_code']
remissions.duplicated(subset=cols_to_check, keep=False).sum()
```

```python
remissions.head()
```

```python
# See rows where typed time, order_date and start_time are duplicated, which could be a sign of duplicates.
typed_time_duplicates = remissions[remissions.duplicated(subset=['typed_time', 'order_date', 'start_time'], keep=False)]
typed_time_duplicates = typed_time_duplicates.sort_values(by='typed_time', ascending=True)

typed_time_duplicates.head(10)

```

```python
typed_time_duplicates['typed_time'].count()
percentage_typed_time_duplicates = (typed_time_duplicates['typed_time'].count() / remissions.shape[0]) * 100
print(f"Percentage of rows with duplicated typed_time, order_date and start_time: {percentage_typed_time_duplicates:.2f}%")
```

None of them where same truck, plant, not order_code, meaning they are not duplicates, only time coincidences that have ocurred less than 300 times throughout >5 years (0.07%)
