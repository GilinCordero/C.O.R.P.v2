```python
# mostrar remisiones al dia, por año con marcadores de cuando empieza cada estacion

for year in sorted(df_total["ticket_date"].dt.year.unique()):
    df_year = df_total[df_total["ticket_date"].dt.year == year]
    remisiones_por_dia_year = (
        df_year.groupby(df_year["ticket_date"].dt.date)
        .size()
        .reset_index(name="cantidad_remisiones")
    )
    remisiones_por_dia_year["ticket_date"] = pd.to_datetime(remisiones_por_dia_year["ticket_date"])
    
    # Calcular medias por estación
    estaciones = {
        "Primavera": (pd.Timestamp(year=int(year), month=3, day=20), pd.Timestamp(year=int(year), month=6, day=20)),
        "Verano": (pd.Timestamp(year=int(year), month=6, day=21), pd.Timestamp(year=int(year), month=9, day=21)),
        "Otoño": (pd.Timestamp(year=int(year), month=9, day=22), pd.Timestamp(year=int(year), month=12, day=20)),
        "Invierno": (pd.Timestamp(year=int(year), month=12, day=21), pd.Timestamp(year=int(year)+1, month=3, day=19))
    }
    
    print(f"\n{'='*60}")
    print(f"Año {year} - Media de remisiones por estación")
    print(f"{'='*60}")
    
    for estacion, (fecha_inicio, fecha_fin) in estaciones.items():
        # Para invierno que cruza años, ajustar para el año actual
        if estacion == "Invierno" and fecha_fin.year > year:
            df_estacion = remisiones_por_dia_year[
                (remisiones_por_dia_year["ticket_date"] >= fecha_inicio) |
                (remisiones_por_dia_year["ticket_date"] <= fecha_fin.replace(year=year))
            ]
        else:
            df_estacion = remisiones_por_dia_year[
                (remisiones_por_dia_year["ticket_date"] >= fecha_inicio) & 
                (remisiones_por_dia_year["ticket_date"] <= fecha_fin)
            ]
        
        if len(df_estacion) > 0:
            media = df_estacion["cantidad_remisiones"].mean()
            std = df_estacion["cantidad_remisiones"].std()
            print(f"{estacion:12} | Media: {media:7.2f} | Desv. Est: {std:7.2f} | Días: {len(df_estacion)}")
        else:
            print(f"{estacion:12} | Sin datos en este año")
    
    fig = px.line(
        remisiones_por_dia_year,
        x="ticket_date",
        y="cantidad_remisiones",
        title=f"Remisiones por día en {year}",
        hover_data={"ticket_date": "|%Y-%m-%d"}
    )
    
    # Agregar marcadores de inicio de estaciones
    estaciones_marcadores = {
        "Primavera": pd.Timestamp(year=int(year), month=3, day=20),
        "Verano": pd.Timestamp(year=int(year), month=6, day=21),
        "Otoño": pd.Timestamp(year=int(year), month=9, day=22),
        "Invierno": pd.Timestamp(year=int(year), month=12, day=21)
    }
    
    for estacion, fecha in estaciones_marcadores.items():
        fig.add_shape(
            type="line",
            x0=fecha,
            x1=fecha,
            y0=0,
            y1=1,
            yref="paper",
            line=dict(color="green", width=2, dash="dash"),
        )
        # Add text annotation manually
        fig.add_annotation(
            x=fecha,
            y=1,
            yref="paper",
            text=estacion,
            showarrow=False,
            xanchor="left",
            yanchor="bottom"
        )
    
    fig.update_layout(xaxis_title="Fecha", yaxis_title="Cantidad de remisiones")
    fig.show()  
```

La media baja considerablemente cada invierno, aunque el invierno pasado no tanto en 2025, algunos veranos tambien parece bajar.  Sobretodo baja en fechas decembrinas

```python
# df con solo los outliers del boxplot anterior

def detectar_outliers_iqr(df, columna):
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    outliers = df[(df[columna] < limite_inferior) | (df[columna] > limite_superior)]
    return outliers

outliers_remisiones_hora = detectar_outliers_iqr(remisiones_hora_510, "remisiones")
print(f"Cantidad de horas outliers: {outliers_remisiones_hora.shape[0]}")
display(outliers_remisiones_hora)
```

```python
# Para las fechas_sin_remision que no sean domingos, añadir una remision por cada una con los datos de promedios moviles como datos de columnas

fechas_sin_remisiones_no_domingo = [
    fecha for fecha in fechas_sin_remisiones if fecha.strftime("%A") != "Sunday"
]
remisiones_a_agregar = []
for fecha in fechas_sin_remisiones_no_domingo:
    fecha_datetime = pd.to_datetime(fecha)
    ticket_hour = fecha_datetime.floor("h")
    
    # Buscar remisiones de esa hora +/- 12 horas
    ventana_inicio = ticket_hour - pd.Timedelta(hours=12)
    ventana_fin = ticket_hour + pd.Timedelta(hours=12)
    
    remisiones_ventana = remisiones_por_hora[
        (remisiones_por_hora["ticket_hour"] >= ventana_inicio) &
        (remisiones_por_hora["ticket_hour"] <= ventana_fin)
    ]
    
    if not remisiones_ventana.empty:
        remision_sintetica = {
            "ticket_id": f"SINTETICO_{fecha}",
            "ticket_date": fecha_datetime,
            "client_name": "SINTETICO",
            "truck_id": "SINTETICO",
            "cantidad_remisiones": remisiones_ventana["cantidad_remisiones"].mean(),
            "media_movil_24h": remisiones_ventana["media_movil_24h"].mean()
        }
        remisiones_a_agregar.append(remision_sintetica)
remisiones_sinteticas_df = pd.DataFrame(remisiones_a_agregar)
print(f"Cantidad de remisiones sintéticas a agregar: {remisiones_sinteticas_df.shape[0]}")
display(remisiones_sinteticas_df.head())
# Crear filas sintéticas compatibles con la estructura de df_510
sinteticas_para_df510 = pd.DataFrame({
    "ticket_date": remisiones_sinteticas_df["ticket_date"] + pd.Timedelta(hours=7),  # dentro de horario laboral
    "order_code": -1,
    "truck_id": -1,
    "plant_id": 510,
    "ticket_quantity": df_510["ticket_quantity"].mean(),
    "typed_time": pd.NaT,
    "at_plant_time": pd.NaT,
    "cycle_time": remisiones_sinteticas_df["media_movil_24h"],
    "client_name": "SINTETICO",
    "ship_address": "SINTETICO",
})

# Recalcular flags igual que en df_510
sinteticas_para_df510["is_holiday"] = sinteticas_para_df510["ticket_date"].dt.date.isin(mexico_holidays)
sinteticas_para_df510["is_christmas_newyear_week"] = (
    (sinteticas_para_df510["ticket_date"].dt.month.eq(12) & sinteticas_para_df510["ticket_date"].dt.day.between(20, 31)) |
    (sinteticas_para_df510["ticket_date"].dt.month.eq(1) & sinteticas_para_df510["ticket_date"].dt.day.between(1, 10))
)
sinteticas_para_df510["is_monday"] = sinteticas_para_df510["ticket_date"].dt.dayofweek.eq(0)
sinteticas_para_df510["is_saturday"] = sinteticas_para_df510["ticket_date"].dt.dayofweek.eq(5)

# (Opcional pero recomendado) bandera de registro sintético
if "is_synthetic" not in df_510.columns:
    df_510["is_synthetic"] = False
sinteticas_para_df510["is_synthetic"] = True

# Alinear columnas y concatenar
for c in df_510.columns:
    if c not in sinteticas_para_df510.columns:
        sinteticas_para_df510[c] = np.nan
sinteticas_para_df510 = sinteticas_para_df510[df_510.columns]

df_510 = pd.concat([df_510, sinteticas_para_df510], ignore_index=True).sort_values("ticket_date").reset_index(drop=True)

print(f"df_510 actualizado: {df_510.shape[0]} filas")
print(f"Filas sintéticas agregadas: {len(df_510[df_510['is_synthetic']])}")
# También agregar sintéticas al df sin outliers
# (usa df_510_no_outliers si existe; crea alias df_510_sin_outliers)

if "df_510_no_outliers" in globals():
    # Evitar duplicados si la celda se corre varias veces
    if "is_synthetic" in df_510_no_outliers.columns:
        base_no_outliers = df_510_no_outliers.loc[~df_510_no_outliers["is_synthetic"]].copy()
    else:
        base_no_outliers = df_510_no_outliers.copy()
        base_no_outliers["is_synthetic"] = False

    sinteticas_no_outliers = sinteticas_para_df510.copy()

    # Alinear columnas contra df_510_no_outliers
    for c in base_no_outliers.columns:
        if c not in sinteticas_no_outliers.columns:
            sinteticas_no_outliers[c] = np.nan
    for c in sinteticas_no_outliers.columns:
        if c not in base_no_outliers.columns:
            base_no_outliers[c] = np.nan

    sinteticas_no_outliers = sinteticas_no_outliers[base_no_outliers.columns]

    df_510_no_outliers = (
        pd.concat([base_no_outliers, sinteticas_no_outliers], ignore_index=True)
        .sort_values("ticket_date")
        .reset_index(drop=True)
    )

    # Alias con el nombre que pediste
    df_510_sin_outliers = df_510_no_outliers

    print(f"df_510_no_outliers actualizado: {df_510_no_outliers.shape[0]} filas")
    print(f"Filas sintéticas en df_510_no_outliers: {df_510_no_outliers['is_synthetic'].sum()}")
else:
    print("No existe df_510_no_outliers en memoria.")
```

Llenamos dias sin remisiones que no fueran domingos con una remision sintética para no perder periodicidad y creamos la variable hour of day para que el modelo pueda aprender de las horas del dia en las que se hacen los pedidos, y no solo de la fecha.

```python
# Pasar lags de las tablas horarias a df_510 y df_510_sin_outliers (celda autosuficiente)
lag_cols = ["lag_1h", "lag_24h", "lag_168h"]


# 2) Reconstruir series horarias y lags para evitar KeyError por orden de ejecución
remisiones_por_hora = (
    df_510.groupby("ticket_date")
    .size()
    .rename("cantidad_remisiones")
    .to_frame()
)

remisiones_por_hora_no_outliers = (
    df_510_no_outliers.groupby("ticket_date")
    .size()
    .rename("cantidad_remisiones")
    .to_frame()
)

for nombre, df_h in [
    ("con_outliers", remisiones_por_hora),
    ("sin_outliers", remisiones_por_hora_no_outliers),
]:
    full_index = pd.date_range(df_h.index.min(), df_h.index.max(), freq="h")
    df_h = df_h.reindex(full_index)
    df_h.index.name = "ticket_date"
    df_h["cantidad_remisiones"] = df_h["cantidad_remisiones"].fillna(0)

    df_h["lag_1h"] = df_h["cantidad_remisiones"].shift(1)
    df_h["lag_24h"] = df_h["cantidad_remisiones"].shift(24)
    df_h["lag_168h"] = df_h["cantidad_remisiones"].shift(168)

    df_h = df_h.reset_index()

    if nombre == "con_outliers":
        remisiones_por_hora = df_h
    else:
        remisiones_por_hora_no_outliers = df_h

# 3) Limpiar lags previos para evitar _x/_y
df_510 = df_510.drop(columns=[c for c in df_510.columns if c.startswith("lag_")], errors="ignore")
df_510_no_outliers = df_510_no_outliers.drop(columns=[c for c in df_510_no_outliers.columns if c.startswith("lag_")], errors="ignore")

# 4) Merge de lags
df_510 = df_510.merge(
    remisiones_por_hora[["ticket_date"] + lag_cols],
    on="ticket_date",
    how="left"
)

df_510_no_outliers = df_510_no_outliers.merge(
    remisiones_por_hora_no_outliers[["ticket_date"] + lag_cols],
    on="ticket_date",
    how="left"
)

print("Lags agregados a ambos DataFrames.")
print("df_510:", [c for c in df_510.columns if c.startswith("lag_")])
print("df_510_no_outliers:", [c for c in df_510_no_outliers.columns if c.startswith("lag_")])
```

```python
# Eliminar filas con NaN en lags dentro de df_510 y df_510_sin_outliers
lag_cols = ["lag_1h", "lag_24h", "lag_168h"]

# Validar que existan columnas lag en ambos DataFrames
faltantes_510 = [c for c in lag_cols if c not in df_510.columns]
faltantes_no_out = [c for c in lag_cols if c not in df_510_no_outliers.columns]

if faltantes_510 or faltantes_no_out:
    print("Faltan columnas de lag para limpiar NaN:")
    print("df_510 faltantes:", faltantes_510)
    print("df_510_no_outliers faltantes:", faltantes_no_out)
else:
    n0_510 = df_510.shape[0]
    n0_no_out = df_510_no_outliers.shape[0]

    df_510 = df_510.dropna(subset=lag_cols).reset_index(drop=True)
    df_510_no_outliers = df_510_no_outliers.dropna(subset=lag_cols).reset_index(drop=True)

    n1_510 = df_510.shape[0]
    n1_no_out = df_510_no_outliers.shape[0]

    print(f"df_510: {n0_510} -> {n1_510} (eliminadas {n0_510 - n1_510} filas)")
    print(f"df_510_no_outliers: {n0_no_out} -> {n1_no_out} (eliminadas {n0_no_out - n1_no_out} filas)")

    display(df_510[["ticket_hour"] + lag_cols].head(10))
    display(df_510_no_outliers[["ticket_hour"] + lag_cols].head(10))
```

```python
# Agregar rolling_mean_24h y rolling_mean_168h a df_510 y df_510_no_outliers
roll_cols = ["rolling_mean_24h", "rolling_mean_168h"]

# Asegurar llave de cruce
df_510["ticket_hour"] = pd.to_datetime(df_510["ticket_date"]).dt.floor("h")
df_510_no_outliers["ticket_hour"] = pd.to_datetime(df_510_no_outliers["ticket_date"]).dt.floor("h")

# Si no existen tablas horarias en memoria, reconstruirlas
if "remisiones_por_hora" not in globals() or "cantidad_remisiones" not in remisiones_por_hora.columns:
    remisiones_por_hora = (
        df_510.groupby("ticket_hour")
        .size()
        .rename("cantidad_remisiones")
        .reset_index()
        .sort_values("ticket_hour")
    )

if "remisiones_por_hora_no_outliers" not in globals() or "cantidad_remisiones" not in remisiones_por_hora_no_outliers.columns:
    remisiones_por_hora_no_outliers = (
        df_510_no_outliers.groupby("ticket_hour")
        .size()
        .rename("cantidad_remisiones")
        .reset_index()
        .sort_values("ticket_hour")
    )

# Calcular rolling means en series horarias
remisiones_por_hora = remisiones_por_hora.sort_values("ticket_hour").copy()
remisiones_por_hora_no_outliers = remisiones_por_hora_no_outliers.sort_values("ticket_hour").copy()

remisiones_por_hora["rolling_mean_24h"] = remisiones_por_hora["cantidad_remisiones"].rolling(window=24, min_periods=1).mean()
remisiones_por_hora["rolling_mean_168h"] = remisiones_por_hora["cantidad_remisiones"].rolling(window=168, min_periods=1).mean()

remisiones_por_hora_no_outliers["rolling_mean_24h"] = remisiones_por_hora_no_outliers["cantidad_remisiones"].rolling(window=24, min_periods=1).mean()
remisiones_por_hora_no_outliers["rolling_mean_168h"] = remisiones_por_hora_no_outliers["cantidad_remisiones"].rolling(window=168, min_periods=1).mean()

# Limpiar rolling previos para evitar _x/_y
df_510 = df_510.drop(columns=[c for c in df_510.columns if c in roll_cols], errors="ignore")
df_510_no_outliers = df_510_no_outliers.drop(columns=[c for c in df_510_no_outliers.columns if c in roll_cols], errors="ignore")

# Merge rolling means a ambos DataFrames
df_510 = df_510.merge(
    remisiones_por_hora[["ticket_hour"] + roll_cols],
    on="ticket_hour",
    how="left"
)

df_510_no_outliers = df_510_no_outliers.merge(
    remisiones_por_hora_no_outliers[["ticket_hour"] + roll_cols],
    on="ticket_hour",
    how="left"
)

print("Rolling means agregados a ambos DataFrames.")
print("df_510:", [c for c in df_510.columns if c in roll_cols])
print("df_510_no_outliers:", [c for c in df_510_no_outliers.columns if c in roll_cols])

display(df_510[["ticket_hour", "rolling_mean_24h", "rolling_mean_168h"]].head(10))
display(df_510_no_outliers[["ticket_hour", "rolling_mean_24h", "rolling_mean_168h"]].head(10))
```
